```python
import numpy as np
import pandas as pd

np.random.seed(42)
```

```python
df = pd.read_csv("../data/raw/loan_risk_dataset.csv")

print("Original Shape:", df.shape)

df.head()
```

```python
columns = [
    "AnnualIncome",
    "Savings",
    "CreditScore",
    "LoanAmount",
    "EmploymentYears",
    "Occupation",
    "Education",
    "PropertyValue",
    "BankBalance"
]
for col in columns:
    missing_index = df.sample(frac = 0.07).index
    df.loc[missing_index,col] = np.nan
```

```python
df.isnull().sum()
```

```python
duplicates = df.sample(frac=0.03, random_state=42)

df = pd.concat([df, duplicates], ignore_index=True)

print(df.shape)
```

```python
income_index = df.sample(frac=0.01).index
df.loc[income_index, "AnnualIncome"] *= 8

saving_index = df.sample(frac=0.01).index
df.loc[saving_index, "Savings"] *= 10

loan_index = df.sample(frac=0.01).index
df.loc[loan_index, "LoanAmount"] *= 5
```

```python
noise = np.random.randint(-2, 3, len(df))

df["Age"] = df["Age"] + noise

df["Age"] = df["Age"].clip(21, 65)
```

```python
invalid_age = df.sample(frac=0.005).index
df.loc[invalid_age, "Age"] = -5

invalid_credit = df.sample(frac=0.005).index
df.loc[invalid_credit, "CreditScore"] = 1200

invalid_loan = df.sample(frac=0.005,random_state=7).index
df.loc[invalid_loan,"LoanAmount"] = -100000
```

```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df.to_csv("../data/processed/loan_risk_dataset_corrupted.csv")
```
