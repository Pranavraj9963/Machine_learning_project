```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("../data/processed/loan_risk_dataset_corrupted.csv")

df.head()
```

```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
df.duplicated().sum()
```

```python
print("Before :",df.shape)

df = df.drop_duplicates()

print("After :",df.shape)
```

```python
df[df["Age"] < 18]
```

```python
median_age = df[df["Age"] >= 18]["Age"].median()

df.loc[df["Age"] < 18,"Age"] = median_age
```

```python
df[df["Age"] < 18]
```

```python
df[df["CreditScore"] > 900]
```

```python
median_credit = df[df["CreditScore"] <= 900]["CreditScore"].median()

df.loc[df["CreditScore"] > 900,"CreditScore"] = median_credit
```

```python
df[df["CreditScore"]> 900]
```

```python
numeric_col = df.select_dtypes(include=["int64","Float64"]).columns
numeric_col
```

```python
imputer = SimpleImputer(strategy="median")
df[numeric_col] = imputer.fit_transform(df[numeric_col])
```

```python
categotical_col = df.select_dtypes(include="object").columns
categotical_col
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent")
df[categotical_col] = cat_imputer.fit_transform(df[categotical_col])
```
