```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/loan_risk_dataset_clean.csv")
df.head()
```

```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
df.shape
```

```python
df["MonthlyIncome"] = df["AnnualIncome"]/12
```

```python
df["DisposableIncome"] = (df["MonthlyIncome"]- df["MonthlyExpenses"])
```

```python
df["EMIRatio"] = (df["ExistingEMI"]/df["MonthlyIncome"])
```

```python
df["SavingRatio"] =(df["Savings"]/df["AnnualIncome"])
```

```python
df["DebtToIncomeRatio"] =(df["LoanAmount"]/df["AnnualIncome"])
```

```python
df["Networth"] = (df["Savings"]+df["Investments"]+df["BankBalance"])
```

```python
df["CreditUtilizationRatio"] =(
    df["CreditCardUsage"] / df["CreditCardLimit"]
)
```

```python
print(df.columns.tolist())
```

```python
df.to_csv("../data/processed/loan_risk_feature_engineered.csv")
```
