```python
import pandas as pd
import joblib

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import AdaBoostClassifier
```

```python
df = pd.read_csv("../data/processed/loan_risk_feature_engineered.csv")
df.head(10)
```

```python
df = df.drop(columns=[
    "CustomerID",
    "FraudFlag",
    "CustomerSegment","Unnamed: 0"
])
```

```python
label = LabelEncoder()

cat_cols = df.select_dtypes(include=["object", "string"]).columns

for col in cat_cols:
    df[col] = label.fit_transform(df[col])

X = df.drop(columns=["LoanApproved"])

y = df["LoanApproved"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = AdaBoostClassifier(
    random_state=42
)

model.fit(
    X_train,
    y_train
)
```

```python
joblib.dump(model,"../models/adaboost_model.pkl")
```

```python
loaded_model = joblib.load("../models/adaboost_model.pkl")

loaded_scaler = joblib.load(
    "../models/scaler.pkl"
)
```
